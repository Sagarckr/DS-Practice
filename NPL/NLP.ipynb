## Text Cleaning and Extraction

```python
import re # regular expression package
```

```python
'''

- re.sub() -> used to clean text.
- re.findall() -> used to extract text.

'''
```

```python
text = "Hello !! sagar@gmail.com    .Store Location :  " \
"Contact : 9833533443 ### ! @ ... jfkds."
```

```python
'''

\d -> digit
\w -> words
\s -> whitespaces
+ -> one or more
* -> zero or more

'''
```

## Clean Text

```python
# re.sub(r'pattern', replace_value, variable)
remove_ws = re.sub(r'\s+', ' ', text)
remove_ws
```

```python
remove_num = re.sub(r'\d+', '', remove_ws)
remove_num
```

```python
remove_wd = re.sub(r'\w+', '', remove_ws)
remove_wd
```

```python
remove_sc = re.sub(r'[!@#..]\s+', '', remove_ws)
remove_sc
```

```python
remove_wc = re.sub(r'[^A-Za-z0-9]\s+', ' ', remove_ws)
remove_wc
```

```python
remove_email = re.sub(r'\w+@\w+.\w+', 'email', remove_ws)
remove_email
```

## Extract Text

```python
extract_email = re.findall(r'\w+@\w+.\w+', remove_ws)
extract_email
```

```python
number = re.findall(r'\d+\s+#', remove_ws)
number
```
