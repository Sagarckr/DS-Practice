```python
'''  
Machine Learning 
-----------------
Type of Machine Learning
-----------------
1. Supervised Learning:
    Feature -> All features must be independent to one another.
    Target -> Target must be dependent to all feature.

    1. Regression
        -> If our target value is continuous use it
    ----------------------------
    Algorithm: 
        1. Linear Regression
        2. Random Forest Regressor
        3. Decision Tree Regressor
        4. XGBoost Regressor
        5. K Nearest Neighbors Regressor
        6. Support Vector Regressor

    2. Classification
        -> If our target value is categorical or discrete then use it.
    ----------------------------
    Algorithm: 
        1. Logistic Regression
        2. Random Forest Classifier
        3. Decision Tree Classifier
        4. XGBoost Classifier
        5. K Nearest Neighbors Classifier
        6. Support Vector Classifier
        
        
2. Unsupervised Learning
    Feature -> All features must be independent to one another.

    1. Clustering
        -> Segmentation
    ----------------------------
    Algorithm:
        1. Agglomerative Clustering
        2. Hierarchical Clustering
        3. K Means Clustering
        4. PCA (Principle Component Analysis)


3. Reinforcement Learning -> Deep Learning
    
'''
```

```python
'''

    Feature Scaling
    ----------------------
    1. Standard Scaler
        -> IF row of data contains outliers or chance of getting outliers.
        -> (Xi - mean)/sd
        -> Best Range (-1 to 1)
        -> (-3 to 3) -> Consider

    2. Min Max Scaler
        -> If row of data does not contains my outliers or change of getting outliers.
        -> (Xi - min)/(max - min)
        -> 0 - 1

    3. Robust Scaler

    4. Feature Scaler

    Feature Encoding
    ----------------------
    1. Label Encoder
    2. One Hot Encoder

    Train Test Rule
    ----------------------
    80/20 Rule -> 80% Train and 20% Test
    70/30
    60/40
    50/50

'''
```

```python
'''

Metrics
-----------------
1. Regression
    -> MAE (Mean Absolute Error) -> sum of all distance / total_data
    -> MSE (Mean Squared Error) -> sum of square distance / total_data
    -> RMSE (Root Mean Squared Error) -> root(MSE) -> Loss function / residual error
    -> R2 Score (Residual Score) -> Accuracy

    -> RMSE Increase - R2 Score Decrease
    -> RMSE Decrease - R2 Score Increase (Correct)

Bo -> Intercept(C)
B1 -> Cofficient / Slope(m)

MLR -> B0 + B1*x1 + B2*x2 + .......... + Bn*xn + e

'''
```
