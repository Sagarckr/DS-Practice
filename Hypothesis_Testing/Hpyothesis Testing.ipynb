## Hypothesis Testing

```python
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv('E:/Courses/DS/Dataset/pizza_sales.csv')
```

## One Sample T-Test
is the average total price of all pizza sales different from $15?

```python
alpha = 0.05
sample = 15
population = df['total_price']
```

```python
t_statistics, p_value = stats.ttest_1samp(population, popmean=sample)
```

```python
print(p_value)
print(t_statistics)
```

```python
ho = f'The average total price of all pizza sales is equal to ${sample}.'
ha = f'The average total price of all pizza sales is not equal to ${sample}.'
```

```python
print(f'Ho: {ho}')
print(f'Ha: {ha}')
```

```python
if p_value < alpha:
    print(f"""Decision: Failed to accept null hypothesis.
Required to accept alternate hypothesis.
Conclusion: {ha}""")
else:
    print(f"""Decision: Failed to reject null hypothesis.
Required to accept null hypothesis.
Conclusion: {ho}""")
```

```python
import numpy as np
actual_mean = population.mean()
sd = population.std()
total_data = np.sqrt(48619)
(actual_mean - sample) / (sd / total_data)
```

## Two Sample T-Test
is the average uniit price of Classic Pizzas and Chicken Pizzas are same?

```python
alpha = 0.05
classic_price = df[df['pizza_category'] == 'Classic']['unit_price']
chicken_price = df[df['pizza_category'] == 'Chicken']['unit_price']
```

```python
# stats.ttest_ind(val1, val2, equal_var = False / True(default))
t_statistics, p_value = stats.ttest_ind(classic_price, chicken_price, equal_var = False)
```

```python
ho = f'The average unit price of classic pizza and chicken are equal.'
ha = f'The average unit price of classic pizza and chicken are not equal.'
```

```python
print(f'Ho: {ho}')
print(f'Ha: {ha}')
```

```python
if p_value < alpha:
    print(f"""Decision: Failed to accept null hypothesis.
Required to accept alternate hypothesis.
Conclusion: {ha}""")
else:
    print(f"""Decision: Failed to reject null hypothesis.
Required to accept null hypothesis.
Conclusion: {ho}""")
```

## ANOVA Test (Analysis of Variance)
is the average total price generated by all four pizza categories (Classic, Chicken, Supreme, and Veggie) differ significantly from one another?

```python
alpha = 0.05
grouped_price = []
data = ['Classic', 'Chicken', 'Supreme', 'Veggie']
for val in data:
    price = df[df['pizza_category'] == val]['total_price']
    grouped_price.append(price)
```

```python
grouped_price
```

```python
t_statistics, p_value = stats.f_oneway(*grouped_price)
```

```python
ho = 'The average total price generated by all four pizza categories are equal.'
ha = 'The average total price generated by all four pizza categories are not equal'
```

```python
print(f'Ho: {ho}')
print(f'Ha: {ha}')
```

```python
if p_value < alpha:
    print(f"""Decision: Failed to accept null hypothesis.
Required to accept alternate hypothesis.
Conclusion: {ha}""")
else:
    print(f"""Decision: Failed to reject null hypothesis.
Required to accept null hypothesis.
Conclusion: {ho}""")
```

## Chi-Squared Test
is pizza category and pizza size dependent to one another?

```python
# crosstab(col1,col2)
contingency = pd.crosstab(df['pizza_category'], df['pizza_size'])
contingency
```

```python
# dof = (number_of_rows - 1) * (number_of_columns - 1)
dof = (4-1) * (5-1)
dof
```

```python
# stats, p_value, degree of freedom, expected frequency
t_statistics, p_value, dof, ef = stats.chi2_contingency(contingency)
```

```python
ho = 'Pizza Category and Pizza size are dependent to each other'
ha = 'Pizza Category and Pizza size are independent to each other'
```

```python
print(f'Ho: {ho}')
print(f'Ha: {ha}')
```

```python
if p_value < alpha:
    print(f"""Decision: Failed to accept null hypothesis.
Required to accept alternate hypothesis.
Conclusion: {ha}""")
else:
    print(f"""Decision: Failed to reject null hypothesis.
Required to accept null hypothesis.
Conclusion: {ho}""")
```
